## Guided Practice: Working with NumPy Arrays

Let's start by creating a NumPy array containing the heights of five plants.

```python
growth = np.array([4,7,5,9,6,8])
```

We can now perform calculations directly on this array.

| Function | What it does |
|:---------|:-------------|
| `np.mean()` | Calculates the average of the values. |
| `np.max()` | Finds the largest value. |
| `np.min()` | Finds the smallest value. |
| `np.sum()` | Calculates the total of all values. |

Let's try these functions:

````python
growth = np.array([4, 7, 5, 9, 6, 8])

average_growth = np.mean(growth)
maximum_growth = np.max(growth)
minimum_growth = np.min(growth)
total_growth = np.sum(growth)

print("Average growth:", average_growth)
print("Maximum growth:", maximum_growth)
print("Minimum growth:", minimum_growth)
print("Total growth:", total_growth)
````

### Boolean Filtering

NumPy also allows us to filter arrays using conditions.

For example, we can find all plants that have a height greater than 12 cm.

````python
# Selecting values greater than 5
tall_plants = growth[growth > 5]

print(tall_plants)
````

### Task 1
Task: Temperature Monitoring

A scientist records the temperature inside a greenhouse over five days.

The recorded temperatures are:

`[22, 25, 21, 28, 24]`

Using NumPy:

- Create an array called `temperatures`.

- Calculate the average temperature.

- Find the highest and lowest temperatures.

- Find all temperatures above 23°C using Boolean filtering.

- Count how many days had temperatures above 23°C.

````python
# Creating an array of the greenhouse temperatures (°C)
temperatures = np.array([22, 25, 21, 28, 24])

# Average, highest and lowest temperatures
average_temperature = np.mean(temperatures)
highest_temperature = np.max(temperatures)
lowest_temperature = np.min(temperatures)

print("Average temperature:", average_temperature)
print("Highest temperature:", highest_temperature)
print("Lowest temperature:", lowest_temperature)

# Boolean filtering: keeping only the temperatures above 23°C
warm_days = temperatures[temperatures > 23]
print("Temperatures above 23°C:", warm_days)

# Counting how many days were above 23°C
number_of_warm_days = len(warm_days)
print("Number of days above 23°C:", number_of_warm_days)
````

## Practice Question: Creating a Research Dataset

A scientist records the following information about five plants:

| Plant | Initial Height (cm) | Final Height (cm) | Sunlight (hours) |
|:------|--------------------:|------------------:|-----------------:|
| Rose  | 10 | 18 | 6 |
| Tulip | 12 | 20 | 7 |
| Lily  | 8 | 15 | 5 |
| Rose  | 11 | 17 | 8 |
| Tulip | 14 | 23 | 6 |

1. Create a dictionary containing the given data.
2. Convert the dictionary into a Pandas DataFrame called `df`.
3. Display the first five rows.
4. Display the shape and column names of the DataFrame.

We can convert this dictionary into a DataFrame using `pd.DataFrame()`.

```python
plants_df = pd.DataFrame(plant_data)
```

Here are some useful Pandas methods and attributes:

| Function / Method | What it does |
|:------------------|:-------------|
| `pd.DataFrame()` | Creates a DataFrame from data. |
| `.head()` | Displays the first five rows. |
| `.shape` | Returns the number of rows and columns. |
| `.columns` | Displays the column names. |

Let's explore our DataFrame:

````python
# guided practice
data = {
    'Plant': ['Rose', 'Tulip', 'Lily', 'Rose', 'Tulip'],
    'Initial_Height': [10, 12, 8, 11, 14],
    'Final_Height': [18, 20, 15, 17, 23],
    'Sunlight': [6, 7, 5, 8, 6]
}

plants_df = pd.DataFrame(data)
# Displaying the first five rows
print(plants_df.head())

# Checking the number of rows and columns
print("Shape:", plants_df.shape)

# Displaying the column names
print("Columns:", plants_df.columns)
````

### Selecting and Filtering Data

We can select a particular column by using its name.

We can also filter rows based on a condition.

For example, let's find all plants with a height greater than 12 cm.

````python
# Note: the DataFrame has no column called 'Height', so we use 'Final_Height'
# (the plant's height at the end of the experiment).

# Selecting the Final_Height column
print(plants_df['Final_Height'])

# Filtering plants with height greater than 12
taller_plants = plants_df[plants_df['Final_Height'] > 12]

print(taller_plants)
````

### Grouping Data

The `groupby()` method allows us to group rows that share the same value.

For example, we can calculate the average height of each plant type.

````python
# Grouping plants by their name and calculating average height
# ('Final_Height' is used because there is no 'Height' column)
average_height = plants_df.groupby('Plant')['Final_Height'].mean()

print(average_height)
````

### Question: **Try and explain this bit of code yourself**
```python
plants_df.groupby('Plant')['Height'].mean()
```

### Answer here:

`plants_df.groupby('Plant')['Height'].mean()` works in three steps:

1. **`plants_df.groupby('Plant')`** splits the rows of the DataFrame into groups based on the value in the `Plant` column. All the Rose rows go into one group, all the Tulip rows into another, and Lily forms a group of its own.
2. **`['Height']`** picks out just the height column inside each group, so only those numbers are used in the calculation. (Our DataFrame doesn't actually have a `Height` column, so in the code above I used `Final_Height` instead.)
3. **`.mean()`** calculates the average of that column for each group separately.

The result is a Pandas Series with one value per plant type, where the plant names are the index:

| Plant | Average final height (cm) |
|:------|--------------------------:|
| Lily  | 15.0 (only one Lily) |
| Rose  | 17.5 = (18 + 17) / 2 |
| Tulip | 21.5 = (20 + 23) / 2 |

In short, it answers the question: *"On average, how tall did each type of plant grow?"*

### Task 2
Continue working with the DataFrame created in the practice question.

- Create a new column called `Growth` by subtracting the initial height from the final height.

- Display only the `Plant` and `Growth` columns.

- Filter and display plants that grew more than 7 cm.

- Calculate the average growth for each plant type using `groupby()`.

````python
# Creating a new Growth column (final height - initial height)
plants_df['Growth'] = plants_df['Final_Height'] - plants_df['Initial_Height']

# Displaying only the Plant and Growth columns
print(plants_df[['Plant', 'Growth']])
print()

# Filtering plants that grew more than 7 cm
fast_growers = plants_df[plants_df['Growth'] > 7]
print("Plants that grew more than 7 cm:")
print(fast_growers)
print()

# Calculating the average growth for each plant type
average_growth_by_type = plants_df.groupby('Plant')['Growth'].mean()
print("Average growth by plant type:")
print(average_growth_by_type)
````

## Part 4 : Combining everything! 
### The School Science Fair

Your school has just held its annual Science Fair! Six students participated, and their marks in Physics, Chemistry and Biology have been recorded.

Your job is to analyse their performance and identify the top-performing students.

### Research Dataset

| Student | Physics | Chemistry | Biology |
|:--------|--------:|-----------:|--------:|
| Asha | 85 | 78 | 92 |
| Jess | 72 | 88 | 80 |
| Meera | 95 | 91 | 89 |
| Rohan | 65 | 70 | 75 |
| Sara | 88 | 84 | 90 |
| Dev | 78 | 76 | 82 |

### Your Mission

**Step 1: Create the Dataset using Pandas**

- Store the information in a dictionary.
- Convert the dictionary into a DataFrame called `students_df`.
- Display the dataset.

**Step 2: Analyse the Scores using NumPy**

- Convert the Physics scores into a NumPy array.
- Calculate the average Physics score.
- Find the highest and lowest Physics scores.

**Step 3: Analyse Student Performance using Pandas**

- Create a new column called `Total` by adding the three subject scores.
- Create another column called `Average` to calculate each student's average marks.
- Filter and display students whose average score is greater than 80.
- Find the student with the highest average score.

**Step 4: Visualise Your Findings using Matplotlib**

- Create a bar graph comparing the average scores of all six students.
- Add a suitable title and label both axes.
- Add a grid to make the graph easier to read.

**Step 5: Answer the Following Questions**

1. Which student scored the highest in Physics?
2. What was the average Physics score?
3. Which student had the highest overall average?
4. How many students scored an average above 80?

````python
# Step 1: Creating the dataset using Pandas

# Storing the marks in a dictionary
student_data = {
    'Student': ['Asha', 'Jess', 'Meera', 'Rohan', 'Sara', 'Dev'],
    'Physics': [85, 72, 95, 65, 88, 78],
    'Chemistry': [78, 88, 91, 70, 84, 76],
    'Biology': [92, 80, 89, 75, 90, 82]
}

# Converting the dictionary into a DataFrame
students_df = pd.DataFrame(student_data)

# Displaying the dataset
print(students_df)
````

````python
# Step 2: Analysing the Physics scores using NumPy

# Converting the Physics column into a NumPy array
physics_scores = np.array(students_df['Physics'])
print("Physics scores:", physics_scores)

# Average, highest and lowest Physics scores
average_physics = np.mean(physics_scores)
highest_physics = np.max(physics_scores)
lowest_physics = np.min(physics_scores)

print("Average Physics score:", average_physics)
print("Highest Physics score:", highest_physics)
print("Lowest Physics score:", lowest_physics)
````

````python
# Step 3: Analysing student performance using Pandas

# Total marks of each student (sum of the three subjects)
students_df['Total'] = students_df['Physics'] + students_df['Chemistry'] + students_df['Biology']

# Average marks of each student (total / 3 subjects), rounded to 2 decimal places
students_df['Average'] = (students_df['Total'] / 3).round(2)

print(students_df)
print()

# Filtering students with an average above 80
above_80 = students_df[students_df['Average'] > 80]
print("Students with an average above 80:")
print(above_80[['Student', 'Average']])
print()

# Finding the student with the highest average score
highest_average = students_df['Average'].max()
top_student = students_df[students_df['Average'] == highest_average]
print("Student with the highest average:")
print(top_student[['Student', 'Average']])
````

````python
# Step 4: Visualising the average scores using Matplotlib

# zorder=3 draws the bars in front of the grid lines
plt.bar(students_df['Student'], students_df['Average'], zorder=3)

# Title and axis labels
plt.title("Average Science Fair Score by Student")
plt.xlabel("Student")
plt.ylabel("Average Score (out of 100)")

# Marks are out of 100, so the y-axis runs from 0 to 100
plt.ylim(0, 100)

# Horizontal grid lines make the bar heights easier to compare
plt.grid(axis='y', alpha=0.4)

plt.show()
````

````python
# Step 5: Working out the answers to the questions

# 1. Student with the highest Physics score
top_physics_student = students_df[students_df['Physics'] == highest_physics]['Student'].iloc[0]
print("1. Highest Physics score:", top_physics_student, "with", highest_physics)

# 2. Average Physics score
print("2. Average Physics score:", average_physics)

# 3. Student with the highest overall average
print("3. Highest overall average:", top_student['Student'].iloc[0], "with", highest_average)

# 4. Number of students with an average above 80
print("4. Students with an average above 80:", len(above_80), "->", list(above_80['Student']))
````

### Step 5: Answers

1. **Which student scored the highest in Physics?** Meera, with 95.
2. **What was the average Physics score?** 80.5
3. **Which student had the highest overall average?** Meera, with an average of 91.67 (total 275).
4. **How many students scored an average above 80?** 3 students: Asha (85.0), Meera (91.67) and Sara (87.33). Jess's average is exactly 80.0, so she is not *above* 80.
